### Q= find the survival rate by age group

```python
# age_bins=np.arange(0,85,5)
# df['age_group']=pd.cut(df['age'],bins=age_bins)
# surv_by_age=df.groupby('age_group',observed=True)['survived'].mean()
# midpoints=[interval.mid for interval in surv_by_age.index]
```

```python
df['age_group']=pd.cut(df['age'],bins=range(0,85,10))
surv_by_age=df.groupby('age_group')['survived'].mean()
plt.figure(figsize=(5,5))
surv_by_age.plot(kind='line',marker='o',figsize=(5,5))
plt.title('survival rate by age')
plt.xlabel('age')
plt.ylabel('survived')
plt.show()
```

### Q. draw a plot to show survival rate according to gender & plot to show count of male & female deathes

```python
df['sex'].value_counts()
```

```python
sc=df['sex'].value_counts()
plt.figure()
c=plt.bar(sc.index,sc.values)
plt.title('count accourding to class')
plt.xlabel('count')
plt.ylabel('sex')
plt.bar_label(c)
plt.show()
```

```python
grp=df.groupby(['class','sex'])['survived'].mean().unstack()
x=np.arange(len(grp.index))
w=0.35

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.bar(x-w/2,grp['male'],w,label='Male')
plt.bar(x+w/2,grp['female'],w,label='Female')
plt.xticks(x,grp.index)
plt.title("survived rate")
plt.legend()

stack=df.groupby(['class','survived']).size().unstack()
plt.subplot(1,2,2)
plt.bar(stack.index,stack[0],label='Died')
plt.bar(stack.index,stack[1],bottom=stack[0],label='survived')
plt.title("outcome counts")
plt.legend()

plt.tight_layout()
plt.show()



```
