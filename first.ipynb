```python
plt.figure(figsize=(5,3))
s=[5,10,15,20,25,30,35,40,45,50]
plt.hist(df['total_bill'],range=(10,40))
plt.title("Distribution of total bill")
plt.xlabel('Total bill')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(5,3))
avg_tip.plot(kind='bar',colormap='gray')
plt.xlabel('days')
plt.ylabel('avg_tip')
plt.title('Distribution of avg tip by days')
```

```python
plt.Figure()
plt.scatter(df['total_bill'],df['tip'],cmap='viridis')
plt.title('tip by bii')
plt.xlabel('total bill')
plt.ylabel('tip')
plt.legend(df['size'])
plt.show()
```

```python
plt.figure()
for size,group in df.groupby('size'):
    count=len(group)
    label_text= f'size{size} (n={count})'
    
    plt.scatter(group['total_bill'],group['tip'],label=label_text)

plt.xlabel('total bill') 
plt.ylabel('Tip')
plt.legend(title="party size and counts")   
```

```python
plt.Figure()
scatter=plt.scatter(df['total_bill'],df['tip'],c=df['size'],cmap='coolwarm')
cbar=plt.colorbar(scatter)
cbar.set_label('party size')
```

```python

avg_tip=df.groupby('day').agg({'tip':'mean'})
avg_tip
plt.figure()
plt.plot(avg_tip,marker='o')
plt.title('abv tip by tip')
plt.xlabel('day')
plt.ylabel('tip')
plt.show()

```

```python
avg_cus=df.groupby('day').agg({'size':'sum'})
avg_cus
plt.figure()
avg_cus.plot(kind='bar')
plt.title('avg coustomer  by day')
plt.xlabel('day')
plt.ylabel('count')
plt.show()
```

```python
cus_sum=df['smoker'].value_counts()
cus_sum
plt.figure(figsize=(3,3))
plt.pie(cus_sum.values,labels=cus_sum.index,autopct='%1.1f%%',shadow=True)
plt.title('smoker vs non-smoker')
plt.show()
```

```python
import numpy as np
x=np.linspace(0,10,100)
fig,ax=plt.subplots(figsize=(5,3.2))
ax.plot(x,np.sin(x),label='sin(x)',color='#2b6cb0')
ax.plot(x,np.cos(x),label='cos(x)',color='#e53e3e',linestyle='--')
ax.set_title('line plot')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
plt.show()


```
